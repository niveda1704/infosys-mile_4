This notebook performs comprehensive Exploratory Data Analysis on the job dataset (`data/jobs.csv`) provided by Intern 2 for the SwipeX AI/ML Career Intelligence Service.

## Objectives:
1. Inspect dataset structure, columns, data types, and quality.
2. Analyze job titles, companies, locations, remote options, job types, experience requirements, salary formats, skills, and descriptions.
3. Check available resume samples in `data/resumes/`.
4. Extract key heuristic features for ATS parsing and job recommendations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import ast
import os

# Set plot styling
plt.style.use('ggplot')
print('Imports successful!')
```

## 1. Load Dataset

```python
data_path = '../data/jobs.csv'
if not os.path.exists(data_path):
    data_path = 'data/jobs.csv'

df = pd.read_csv(data_path)
print(f'Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

## 2. Dataset Overview & Data Quality Checks

```python
print('--- Dataset Info ---')
print(df.info())

print('\n--- Missing Values ---')
print(df.isnull().sum())

print(f'\nDuplicate Rows: {df.duplicated().sum()}')
```

## 3. Job Role Analysis

```python
job_titles = df['Job Title'].value_counts()
print(f'Unique Job Titles ({len(job_titles)}):')
print(job_titles)

# Visualization
plt.figure(figsize=(10, 6))
job_titles.plot(kind='barh', color='skyblue')
plt.title('Job Title Distribution')
plt.xlabel('Count')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 4. Company & Company Type Analysis

```python
print('--- Company Type Distribution ---')
company_types = df['Company Type'].value_counts()
print(company_types)

print('\n--- Top Companies ---')
print(df['Company'].value_counts().head(10))

plt.figure(figsize=(6, 4))
company_types.plot(kind='pie', autopct='%1.1f%%', colors=['#66b3ff', '#ff9999'])
plt.title('Company Type Breakdown (MNC vs Startup)')
plt.ylabel('')
plt.show()
```

## 5. Location & Remote Job Analysis

```python
print('--- Location Distribution ---')
print(df['Location'].value_counts())

print('\n--- Remote Option Distribution ---')
remote_dist = df['Remote'].value_counts()
print(remote_dist)

plt.figure(figsize=(8, 4))
df['Location'].value_counts().plot(kind='bar', color='teal')
plt.title('Job Location Distribution')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Job Type Analysis

```python
job_types = df['Job Type'].value_counts()
print('Job Types:')
print(job_types)

plt.figure(figsize=(5, 4))
job_types.plot(kind='bar', color='coral')
plt.title('Job Type Breakdown')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 7. Experience Requirements Analysis

```python
exp_dist = df['Experience'].value_counts()
print('Experience Level Breakdown:')
print(exp_dist)

plt.figure(figsize=(6, 4))
exp_dist.plot(kind='bar', color='purple')
plt.title('Experience Requirement Distribution')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 8. Salary Analysis

```python
salary_dist = df['Salary'].value_counts()
print('Salary Range Distribution:')
print(salary_dist)

# Observations on Salary
print('\nObservation: Salary is stored as range strings (e.g. "3-5 LPA", "8-12 LPA").')
```

## 9. Required Skills Analysis

```python
# Parse skill string lists
all_skills = []
for item in df['Required Skills'].dropna():
    try:
        parsed = ast.literal_eval(item)
        if isinstance(parsed, list):
            all_skills.extend([s.strip() for s in parsed])
        else:
            all_skills.append(str(parsed).strip())
    except:
        all_skills.extend([s.strip() for s in str(item).split(',')])

skill_counts = pd.Series(all_skills).value_counts()
print(f'Total Skill Occurrences: {len(all_skills)}')
print(f'Unique Skills Identified: {len(skill_counts)}')
print('\nTop 20 Most Demanded Skills:')
print(skill_counts.head(20))

plt.figure(figsize=(10, 6))
skill_counts.head(15).plot(kind='barh', color='seagreen')
plt.title('Top 15 Most Demanded Skills Across Jobs')
plt.xlabel('Frequency')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 10. Job Description Analysis

```python
char_lens = df['Job Description'].apply(len)
word_counts = df['Job Description'].apply(lambda x: len(x.split()))

print('--- Job Description Character Statistics ---')
print(f'Min: {char_lens.min()}, Max: {char_lens.max()}, Mean: {char_lens.mean():.2f}, Median: {char_lens.median()}')

print('\n--- Job Description Word Count Statistics ---')
print(f'Min: {word_counts.min()}, Max: {word_counts.max()}, Mean: {word_counts.mean():.2f}, Median: {word_counts.median()}')

print('\n--- Sample Job Descriptions ---')
for idx, desc in enumerate(df['Job Description'].head(3)):
    print(f'Sample {idx+1}: {desc}')
```

## 11. Resume Sample Analysis

```python
resume_dir = '../data/resumes/'
if not os.path.exists(resume_dir):
    resume_dir = 'data/resumes/'

resume_files = [f for f in os.listdir(resume_dir) if not f.startswith('.')]
print(f'Available Resume Files in data/resumes/: {len(resume_files)}')
if len(resume_files) == 0:
    print('Status: Resume EDA is PENDING/BLOCKED as no sample resume files exist in data/resumes/.')
```
